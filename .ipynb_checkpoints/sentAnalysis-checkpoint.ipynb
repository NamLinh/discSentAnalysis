```python
# library imports

import pandas as pd
import nltk
import numpy as np
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
# Defining the preprocessing for text
def preprocess_text(text):

    # Tokenising:
    tokens = word_tokenize(str(text).lower())

    # Remove stop words, i.e. words that generally don't add any emotion to sentences, e.g. and, but, etc.

    filtered_tokens = [token for token in tokens if (token not in stopwords.words('english') or token not in "0123456789")]

    # Lemmatize the tokens, i.e. converting all words into their base form. For example, "gaming" -> game, "hiking" -> hike, and so on.

    lemmatizer = WordNetLemmatizer()
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]

    # Merging all the tokens back into one string

    processed_text = ' '.join(lemmatized_tokens)

    return processed_text

# Applying the function to our data.

#df['reviewText'] = df['reviewText'].apply(preprocess_text)
#df.head(5)
```

```python
# Creating the NLTK sentiment analyser.

analyzer = SentimentIntensityAnalyzer()


# Defining the sentiment retrieval function which just utilises the NLTK library 
# to output the overall sentiment for our text input parameter.

def get_sentiment(text):

    scores = analyzer.polarity_scores(text)

    sentiment = 1 if scores['pos'] > 0 else 0

    return sentiment

# Applying the same function to our dataset.

#df['sentiment'] = df['reviewText'].apply(get_sentiment)
#df.head(5)
```

```python
# Now going to use my friends' messages.

disc_df = pd.read_csv("messages2.csv")
disc_df.head(5)
```

```python
# Removing a bunch of N/A, and other such irrelevant data from the messages column, and in turn, the records. Shouldn't matter too much as there are so
# many messages that it shouldn't make a significant enough impact.

rem_indexes = []
for i in range(len(disc_df["content"])):
    rec = disc_df["content"][i]
    if pd.isna(rec) or "http" in rec:
        rem_indexes.append(i)

if rem_indexes:
    disc_df = disc_df.drop(rem_indexes)
disc_df.head(5)
```

```python
rem_author_indexes = []
for i in range(len(disc_df["author"])):
    rec = disc_df["author"][i]
    if rec == "jesona":
        rem_author_indexes.append(i)

if rem_author_indexes:
    disc_df = disc_df.drop(rem_author_indexes)
disc_df.head(5)
```

```python
# Preprocessing the text data, i.e. tokenising, lemmatising, etc.

disc_df['content'] = disc_df['content'].apply(preprocess_text)
disc_df.head(5)
```

```python
disc_df['sentiment'] = disc_df['content'].apply(get_sentiment)
disc_df.head(5)
```

```python
pos_neg_dico = {"pos":0, "neg":0}
for rec in disc_df['sentiment']:
    if int(rec) == 0:
        pos_neg_dico["neg"] += 1
    else:
        pos_neg_dico["pos"] += 1

disc_pos_score = round((pos_neg_dico["pos"] / (pos_neg_dico["pos"] + pos_neg_dico["neg"])) * 100, 2)

print(f"Your friendship group's positivity score is {disc_pos_score}%!")
```

```python
pos_neg_by_author_dico = {}

for i in range(len(disc_df)):
    author = disc_df['author'].iloc[i]
    sent = disc_df['sentiment'].iloc[i]
    if author not in pos_neg_by_author_dico:
        pos_neg_by_author_dico[author] = {"pos":0, "neg":0}
    else:
        if sent == 0:
            pos_neg_by_author_dico[author]["neg"] += 1
        else:
            pos_neg_by_author_dico[author]["pos"] += 1

pos_neg_by_author_dico
```

```python
for author in pos_neg_by_author_dico.keys():
    temp = pos_neg_by_author_dico[author]
    print(f"{author} has a positivity score of {round(temp['pos'] / (temp['pos'] + temp['neg']) * 100, 2)}%")
```

```python
# Now we have a look into the actual results of what was assigned positive labels or negative labels, and choose samples at random. 
# Interestingly, we see messages such as just simple emojis being given a negative score. Some otherwise "neutral statements" 
# were also not the most accurate

import random

rand_list = [random.randint(1,500)*x for x in range(1,6)]

for elem in rand_list:
    print(f"{disc_df['author'].iloc[elem]} said \"{disc_df['content'].iloc[elem]}\" and that was assigned a positivity score of {disc_df['sentiment'].iloc[elem]}")
```
